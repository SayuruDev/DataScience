```python
!pip install pyspark
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("PySpark RDD Lesson") \
    .getOrCreate()
sc = spark.sparkContext
```

```python
data = [1, 2, 3, 4, 5]
rdd = sc.parallelize(data)
print("RDD Contents:", rdd.collect())
```

```python
rdd_file = sc.textFile("banknote_data.txt")
print("First 5 lines:", rdd_file.take(5))
```

```python
rdd.saveAsTextFile("bankdata_save.txt")
```

```python
lookup = {"A": 1, "B": 2, "C": 3}
broadcast_var = sc.broadcast(lookup)

rdd_keys = sc.parallelize(["A", "B", "C", "D"])
mapped = rdd_keys.map(lambda x: broadcast_var.value.get(x, 0))
print("Mapped Values:", mapped.collect())
```

```python
accum = sc.accumulator(0)

def count_even(x):
    global accum
    if x % 2 == 0:
        accum += 1

rdd_nums = sc.parallelize([1, 2, 3, 4, 5, 6])
rdd_nums.foreach(count_even)
print("Even Number Count:", accum.value)
```
